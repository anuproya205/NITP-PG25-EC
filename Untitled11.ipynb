```python
import numpy as np
import pandas as pd

# ============================================================
# SOIL MOISTURE RETRIEVAL USING MODIFIED DUBOIS + WCM
# FDP LUT DATA
# ============================================================


# ============================================================
# 1. OBSERVED SENTINEL-1 VV
# ============================================================

sigma_vv_obs = 0.0287

print("Observed VV =", sigma_vv_obs)


# ============================================================
# 2. RADAR / SOIL PARAMETERS
# ============================================================

theta_deg = 40
lambda_cm = 5.6
s_cm = 1.0

theta = np.deg2rad(theta_deg)

cos_theta = np.cos(theta)
sin_theta = np.sin(theta)

sec_theta = 1 / cos_theta
cot_theta = 1 / np.tan(theta)

# Radar wavenumber
k = 2 * np.pi / lambda_cm

print("Incidence angle =", theta_deg, "degree")
print("Wavelength      =", lambda_cm, "cm")
print("RMS roughness   =", s_cm, "cm")
print("Wavenumber k    =", k)


# ============================================================
# 3. FDP LOOK-UP TABLE
#
# LAI and SM MUST remain paired row-by-row
# ============================================================

lut = pd.DataFrame({

    "LAI": [
        7.49,
        7.57,
        7.78,
        7.12,
        7.22,
        6.15,
        7.07,
        4.58,
        6.11,
        8.77,
        8.93,
        8.67,
        8.80,
        8.26,
        8.33,
        6.34,
        7.14,
        7.47,
        5.66,
        5.16,
        6.22
    ],

    "SM": [
        0.4694,
        0.4694,
        0.4694,
        0.2219,
        0.2219,
        0.2219,
        0.4417,
        0.4417,
        0.4417,
        0.3413,
        0.3413,
        0.3413,
        0.4375,
        0.4375,
        0.4375,
        0.4184,
        0.4184,
        0.4184,
        0.436666667,
        0.441666667,
        0.435333333
    ],

    # These are the VV and VH values printed in the FDP slide.
    # They are kept here for reference/verification.
    "VV_FDP": [
        0.03859,
        0.04903,
        0.03493,
        0.02832,
        0.02317,
        0.03057,
        0.04429,
        0.02798,
        0.03680,
        0.05130,
        0.04578,
        0.04100,
        0.04121,
        0.04840,
        0.04711,
        0.05173,
        0.03546,
        0.04658,
        0.05107,
        0.05242,
        0.04297
    ],

    "VH_FDP": [
        0.02793,
        0.01725,
        0.02207,
        0.01524,
        0.01381,
        0.01683,
        0.02428,
        0.01586,
        0.02334,
        0.02434,
        0.02548,
        0.01834,
        0.01996,
        0.02899,
        0.02539,
        0.02040,
        0.02671,
        0.03059,
        0.02360,
        0.03759,
        0.03194
    ]
})


# ============================================================
# 4. MODIFIED DUBOIS MODEL
#
# sigma_VV_soil =
# 10^(-1.138)
# * cos(theta)^1.528
# * 10^(0.008*cot(theta)*SM)
# * (k*s)^(0.71*sin(theta))
# ============================================================

def modified_dubois_vv(SM):

    sigma_soil_vv = (
        10**(-1.138)
        * cos_theta**1.528
        * 10**(
            0.008
            * cot_theta
            * SM
        )
        * (k * s_cm)**(
            0.71 * sin_theta
        )
    )

    return sigma_soil_vv


# Calculate soil VV for every LUT row
lut["sigma_soil_VV"] = lut["SM"].apply(
    modified_dubois_vv
)


# ============================================================
# 5. WCM A AND B SEARCH RANGE
# ============================================================
#
# IMPORTANT:
# Your sir wrote:
# "take A & B range according to the paper which uses LAI"
#
# Therefore these are PARAMETERS TO BE CONFIRMED from that paper.
#
# The values below are only a computational example.
#
# For winter wheat, published WCM literature has used
# A = 0.0018 and B = 0.138.
#
# If your sir has specified another paper/range,
# replace these ranges accordingly.
# ============================================================

A_min = 0.001
A_max = 0.003
A_step = 0.00001

B_min = 0.10
B_max = 0.20
B_step = 0.001


A_values = np.arange(
    A_min,
    A_max + A_step,
    A_step
)

B_values = np.arange(
    B_min,
    B_max + B_step,
    B_step
)


# ============================================================
# 6. WCM CALCULATION
# ============================================================

results = []


for row_number, row in lut.iterrows():

    # --------------------------------------------------------
    # Take LAI and SM from SAME LUT row
    # --------------------------------------------------------

    LAI = row["LAI"]
    SM = row["SM"]

    # Soil backscatter calculated from this row's SM
    sigma_soil_vv = row["sigma_soil_VV"]


    # --------------------------------------------------------
    # Supervisor instructed:
    #
    # V1 = V2 = LAI
    # --------------------------------------------------------

    V1 = LAI
    V2 = LAI


    # --------------------------------------------------------
    # Search A
    # --------------------------------------------------------

    for A in A_values:


        # ----------------------------------------------------
        # Search B
        # ----------------------------------------------------

        for B in B_values:


            # =================================================
            # WCM EQUATION 1
            #
            # tau^2 =
            # exp(-2 B V2 sec(theta))
            # =================================================

            tau_squared = np.exp(
                -2
                * B
                * V2
                * sec_theta
            )


            # =================================================
            # WCM VEGETATION BACKSCATTER
            #
            # sigma_veg =
            # A V1 cos(theta)
            # (1 - tau^2)
            # =================================================

            sigma_veg = (
                A
                * V1
                * cos_theta
                * (1 - tau_squared)
            )


            # =================================================
            # TOTAL WCM BACKSCATTER
            #
            # sigma_total =
            # sigma_veg +
            # tau^2 * sigma_soil
            # =================================================

            sigma_vv_total = (
                sigma_veg
                + tau_squared * sigma_soil_vv
            )


            # =================================================
            # ERROR
            #
            # Only observed VV is being used in THIS WCM step.
            # =================================================

            difference = (
                sigma_vv_obs
                - sigma_vv_total
            )


            # For one observed VV value:
            # RMSE = absolute error

            cost = np.sqrt(
                difference**2
            )


            # Store result
            results.append([
                row_number + 1,
                LAI,
                SM,
                sigma_soil_vv,
                A,
                B,
                tau_squared,
                sigma_veg,
                sigma_vv_total,
                difference,
                cost
            ])


# ============================================================
# 7. CREATE FINAL WCM LUT
# ============================================================

columns = [
    "LUT_Row",
    "LAI",
    "SM",
    "sigma_soil_VV",
    "A",
    "B",
    "tau_squared",
    "sigma_veg",
    "sigma_vv_total",
    "difference",
    "cost"
]

wcm_lut = pd.DataFrame(
    results,
    columns=columns
)


# ============================================================
# 8. FIND MINIMUM COST
# ============================================================

best_index = wcm_lut["cost"].idxmin()

best_result = wcm_lut.loc[
    best_index
]


# ============================================================
# 9. PRINT BEST RESULT
# ============================================================

print("\n")
print("=" * 70)
print("                 BEST WCM RESULT")
print("=" * 70)

print(
    f"Observed VV             = "
    f"{sigma_vv_obs:.15f}"
)

print(
    f"LUT Row                 = "
    f"{int(best_result['LUT_Row'])}"
)

print(
    f"Best LAI                = "
    f"{best_result['LAI']:.6f}"
)

print(
    f"Best SM                 = "
    f"{best_result['SM']:.9f}"
)

print(
    f"Soil VV from Dubois     = "
    f"{best_result['sigma_soil_VV']:.15f}"
)

print(
    f"Best A                  = "
    f"{best_result['A']:.8f}"
)

print(
    f"Best B                  = "
    f"{best_result['B']:.8f}"
)

print(
    f"Tau squared             = "
    f"{best_result['tau_squared']:.15f}"
)

print(
    f"Vegetation VV           = "
    f"{best_result['sigma_veg']:.15f}"
)

print(
    f"Calculated Total VV     = "
    f"{best_result['sigma_vv_total']:.15f}"
)

print(
    f"Difference              = "
    f"{best_result['difference']:.15f}"
)

print(
    f"Minimum Cost            = "
    f"{best_result['cost']:.15f}"
)


# ============================================================
# 10. DISPLAY COMPLETE 21-ROW LUT
# ============================================================

print("\n")
print("=" * 70)
print("          21 FDP LUT ROWS WITH DUBOIS SOIL VV")
print("=" * 70)

print(
    lut[
        [
            "LAI",
            "SM",
            "sigma_soil_VV",
            "VH_FDP",
            "VV_FDP"
        ]
    ].to_string(
        index=True,
        float_format=lambda x: f"{x:.9f}"
    )
)


# ============================================================
# 11. DISPLAY 20 LOWEST-COST WCM COMBINATIONS
# ============================================================

best_20 = (
    wcm_lut
    .sort_values(
        by="cost",
        ascending=True
    )
    .head(20)
)


print("\n")
print("=" * 70)
print("          20 LOWEST-COST WCM COMBINATIONS")
print("=" * 70)

print(
    best_20[
        [
            "LUT_Row",
            "LAI",
            "SM",
            "A",
            "B",
            "tau_squared",
            "sigma_soil_VV",
            "sigma_veg",
            "sigma_vv_total",
            "difference",
            "cost"
        ]
    ].to_string(
        index=False,
        float_format=lambda x: f"{x:.12f}"
    )
)
```
