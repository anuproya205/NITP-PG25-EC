```python
import math
import pandas as pd

# --- Coefficients alag-alag ---
A_VV, A_VH = 0.08, 0.03  # VV ke liye bada, VH ke liye chota - paper se change karna
B, C, D = 0.02, 15.0, 0.14
E, F = 0.9, 0.8
theta = math.radians(30)

LUT = []
for LAI in [0, 1, 2, 3, 4, 5]:
    for SM in [0.25]:
        W = LAI
        if LAI == 0:
            tau2 = 1.0
            sigma_veg_VV = 0.0
            sigma_veg_VH = 0.0
        else:
            tau2 = math.exp(-2*B*(W**F)/math.cos(theta))
            sigma_veg_VV = A_VV * (LAI**E) * math.cos(theta) * (1 - tau2)
            sigma_veg_VH = A_VH * (LAI**E) * math.cos(theta) * (1 - tau2)

        sigma_soil = D * math.exp(C * SM) * tau2 * math.cos(theta)

        sigma_total_VV = sigma_veg_VV + sigma_soil
        sigma_total_VH = sigma_veg_VH + sigma_soil

        LUT.append([sigma_total_VV, sigma_total_VH, LAI, SM])

df = pd.DataFrame(LUT, columns=['VV_LUT', 'VH_LUT', 'LAI', 'SM'])
print(df)
df.to_csv('LUT_table_final.csv', index=False)
```
