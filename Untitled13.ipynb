```python
 These are DEMONSTRATION values, not measured wheat parameters.
#
# vwc_kg_m2: vegetation water content
# height_m: approximate canopy height
# cover: fractional canopy cover, between 0 and 1
# leaf_fraction, stem_fraction, ear_fraction:
# simplified structural fractions, not physical DBA inputs by themselves
# ieee paper refer
wheat_stages = {

     #Stage 1: Just after emergence, mostly soil visible
    "Early growth": {
        "vwc_kg_m2": 0.15,  # Very low water content, plant is tiny
        "height_m": 0.10,   # 10 cm height
        "cover": 0.15,   # Only 15% ground covered, 85% soil visible
        "leaf_fraction": 0.80,  # 80% of plant is leaves
        "stem_fraction": 0.20,  # 20% is stem
        "ear_fraction": 0.00   # No ear yet
    },

      # Stage 2: Leaf production stage
    "Tillering": {
        "vwc_kg_m2": 0.50, # VWC increases as biomass grows
        "height_m": 0.25,
        "cover": 0.40, # 40% cover, leaves spreading
        "leaf_fraction": 0.70,  # Leaves still dominant but less than before
        "stem_fraction": 0.30,   # Stem contribution increases
        "ear_fraction": 0.00  # Still no ear
    },

     # Stage 3: Vertical growth stage
    "Stem elongation": {
        "vwc_kg_m2": 1.20,  # Rapid water uptake
        "height_m": 0.55,  # 55 cm, stem elongates fast
        "cover": 0.70,   # 70% canopy closure
        "leaf_fraction": 0.50,   #50% contribustion now leaf and stem both are equal
        "stem_fraction": 0.50, # Stem becomes 50%
        "ear_fraction": 0.00   # Ear not emerged yet
    },

    # Stage 4: Reproductive stage begins
    "Heading/flowering": {
        "vwc_kg_m2": 2.00,  # Peak vegetative water content
        "height_m": 0.80,   # Near max height 80 cm
        "cover": 0.90,   # 90% closed canopy
        "leaf_fraction": 0.40, # Leaf fraction starts decreasing
        "stem_fraction": 0.40,   # Stem stable
        "ear_fraction": 0.20  # Ear appears - 20% of structure
    },
     # Stage 5: Grain develops
    "Grain filling": {
        "vwc_kg_m2": 2.20,      # Max VWC, highest biomass + water
        "height_m": 0.85,       # Max height 85 cm
        "cover": 0.95,          # Almost full cover
        "leaf_fraction": 0.30,  # Leaves senesce, drop to 30%
        "stem_fraction": 0.40,  # Stem still 40%
        "ear_fraction": 0.30    # Ear becomes 30% - grain heavy
    },

    # where from this value is taken to identify this - This is DEMO value, not measured. Taken from literature trend
    # Stage 6: Senescence - plant dries
    "Maturity": {
        "vwc_kg_m2": 0.80,      # VWC drops sharply - plant dries
        "height_m": 0.80,       # Height slightly shrinks due to bending/drying
        "cover": 0.90,          # Cover stays high but leaves dry
        "leaf_fraction": 0.25,  # Leaf minimum - mostly dead
        "stem_fraction": 0.45,  # Stem becomes dominant - dry stalk
        "ear_fraction": 0.30    # Ear remains 30% - dry grain
    }
}

# Convert dictionary to pandas DataFrame and transpose (.T) so stages become rows
stage_table = pd.DataFrame(wheat_stages).T

# Display the table in notebook (Jupyter / Colab)
display(stage_table)
```
