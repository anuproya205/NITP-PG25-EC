```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# =====================================================
# 1. CONSTANTS AND INPUT PARAMETERS
# =====================================================

EPSILON_0 = 8.8541878128e-12

# Sentinel-1 C-band frequency
frequency_GHz = 5.405
frequency = frequency_GHz * 1e9

# Temporary sample soil parameters
sand_percent = 40
clay_percent = 20
bulk_density = 1.40
temperature_C = 25

# Convert percentages into fractions
S = sand_percent / 100
C = clay_percent / 100

# =====================================================
# 2. DOBSON EMPIRICAL COEFFICIENTS
# =====================================================

alpha = 0.65

beta1 = 1.27 - 0.519*S - 0.152*C
beta2 = 2.06 - 0.928*S - 0.255*C

# =====================================================
# 3. EFFECTIVE SOIL CONDUCTIVITY
# =====================================================

sigma_s = (
    -1.645
    + 1.939*bulk_density
    - 2.256*S
    + 1.594*C
)

if sigma_s < 0:
    raise ValueError("Negative effective conductivity. Check input parameters.")

# =====================================================
# 4. PURE WATER DIELECTRIC PROPERTIES
# =====================================================

epsilon_water_infinity = 4.9

epsilon_water_static = (
    88.045
    - 0.4147*temperature_C
    + 6.295e-4*temperature_C**2
    + 1.075e-5*temperature_C**3
)

tau_water = (
    1.1109e-10
    - 3.824e-12*temperature_C
    + 6.938e-14*temperature_C**2
    - 5.096e-16*temperature_C**3
) / (2*np.pi)

x = 2*np.pi*frequency*tau_water

epsilon_water_real = (
    epsilon_water_infinity
    + (epsilon_water_static - epsilon_water_infinity)/(1+x**2)
)

epsilon_water_imag = (
    x*(epsilon_water_static - epsilon_water_infinity)/(1+x**2)
)

# =====================================================
# 5. DOBSON MODEL FUNCTION
# =====================================================

def dobson_model(mv):

    if not (0 < mv <= 1):
        raise ValueError("Soil moisture must be between 0 and 1.")

    # Conductivity contribution
    conductivity_loss = (
        (2.65 - bulk_density)/(2.65*mv)
        * sigma_s/(2*np.pi*EPSILON_0*frequency)
    )

    epsilon_water_imag_total = (
        epsilon_water_imag + conductivity_loss
    )

    # Real part of soil dielectric constant
    epsilon_real = (
        1 + 0.66*bulk_density
        + mv**beta1 * epsilon_water_real**alpha
        - mv
    )**(1/alpha)

    # Imaginary part of soil dielectric constant
    epsilon_imag = (
        mv**beta2 * epsilon_water_imag_total
    )

    # Complex dielectric constant
    epsilon_complex = complex(epsilon_real, -epsilon_imag)

    return epsilon_real, epsilon_imag, epsilon_complex


# =====================================================
# 6. SOIL MOISTURE RANGE
# =====================================================

soil_moisture_values = np.arange(0.05, 0.401, 0.05)

# =====================================================
# 7. RUN DOBSON MODEL
# =====================================================

results = []

for mv in soil_moisture_values:

    epsilon_real, epsilon_imag, epsilon_complex = dobson_model(mv)

    results.append({
        "Soil Moisture": mv,
        "Epsilon Real": epsilon_real,
        "Epsilon Imaginary": epsilon_imag,
        "Complex Dielectric": epsilon_complex
    })

df = pd.DataFrame(results)

# =====================================================
# 8. DISPLAY RESULTS
# =====================================================

print("DOBSON MODEL RESULTS")
print("====================")
print("Frequency:", frequency_GHz, "GHz")
print("Sand:", sand_percent, "%")
print("Clay:", clay_percent, "%")
print("Bulk density:", bulk_density, "g/cm3")
print("Temperature:", temperature_C, "°C")
print("Beta 1:", beta1)
print("Beta 2:", beta2)
print("Effective conductivity:", sigma_s)
print()

display(df.round(5))

# =====================================================
# 9. GRAPH 1: SOIL MOISTURE VS REAL DIELECTRIC
# =====================================================

plt.figure(figsize=(8,5))

plt.plot(
    df["Soil Moisture"],
    df["Epsilon Real"],
    marker="o"
)

plt.xlabel("Volumetric Soil Moisture (m³/m³)")
plt.ylabel("Real Dielectric Constant (ε')")
plt.title("Soil Moisture vs Real Dielectric Constant")
plt.grid(True)
plt.show()

# =====================================================
# 10. GRAPH 2: SOIL MOISTURE VS IMAGINARY DIELECTRIC
# =====================================================

plt.figure(figsize=(8,5))

plt.plot(
    df["Soil Moisture"],
    df["Epsilon Imaginary"],
    marker="o"
)

plt.xlabel("Volumetric Soil Moisture (m³/m³)")
plt.ylabel("Imaginary Dielectric Constant (ε'')")
plt.title("Soil Moisture vs Imaginary Dielectric Constant")
plt.grid(True)
plt.show()

# =====================================================
# 11. GRAPH 3: BOTH DIELECTRIC COMPONENTS
# =====================================================

plt.figure(figsize=(8,5))

plt.plot(
    df["Soil Moisture"],
    df["Epsilon Real"],
    marker="o",
    label="Real part (ε')"
)

plt.plot(
    df["Soil Moisture"],
    df["Epsilon Imaginary"],
    marker="s",
    label="Imaginary part (ε'')"
)

plt.xlabel("Volumetric Soil Moisture (m³/m³)")
plt.ylabel("Dielectric Constant")
plt.title("Dobson Model: Dielectric Response to Soil Moisture")
plt.legend()
plt.grid(True)
plt.show()

# =====================================================
# 12. SAVE RESULTS
# =====================================================

df.to_csv("Dobson_Soil_Moisture_Results.csv", index=False)

print("Results saved as Dobson_Soil_Moisture_Results.csv")
print("Dobson sensitivity analysis completed successfully!")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# =====================================================
# MODULE 2: FRESNEL REFLECTION COEFFICIENT
# Using complex dielectric values from Module 1
# =====================================================

# Temporary incidence angle for testing
theta_deg = 40
theta = np.deg2rad(theta_deg)

# Extract complex dielectric constant from Module 1
epsilon_r = np.asarray(df["Complex Dielectric"], dtype=complex)

# Calculate the square-root term
root_term = np.sqrt(epsilon_r - np.sin(theta)**2 + 0j)

# VV (vertical / p-polarization) Fresnel reflection coefficient
r_vv = (
    epsilon_r * np.cos(theta) - root_term
) / (
    epsilon_r * np.cos(theta) + root_term
)

# Power reflectance at a smooth air-soil interface
R_vv = np.abs(r_vv)**2

# Add results to a copy of the Module 1 table
module2_df = df.copy()
module2_df["Fresnel r_VV"] = r_vv
module2_df["VV Power Reflectance"] = R_vv

# Display results
print("MODULE 2: FRESNEL REFLECTION RESULTS")
print("Incidence angle:", theta_deg, "degrees")
display(module2_df)

# Plot 1: Soil moisture vs VV power reflectance
moisture_col = next(
    (c for c in ["mv", "Soil Moisture", "Soil Moisture (mv)",
                 "Volumetric Soil Moisture"]
     if c in module2_df.columns),
    None
)

if moisture_col is not None:
    x = module2_df[moisture_col]
    x_label = "Volumetric soil moisture (mv)"
else:
    x = np.arange(len(module2_df))
    x_label = "Sample number"

plt.figure(figsize=(7, 5))
plt.plot(x, R_vv, marker="o")
plt.xlabel(x_label)
plt.ylabel("VV power reflectance")
plt.title("Fresnel VV Power Reflectance vs Soil Moisture")
plt.grid(True)
plt.show()

# Save results
module2_df.to_csv("Module2_Fresnel_Results.csv", index=False)
print("Results saved as Module2_Fresnel_Results.csv")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# =====================================================
# MODULE 3: SOIL ROUGHNESS AUTOCORRELATION + SPECTRUM
# Provisional exponential correlation model
# =====================================================

# 1. Sentinel-1 test parameters
frequency = 5.405e9       # Hz
theta_deg = 40            # temporary incidence angle
s = 0.01                  # RMS roughness: 1 cm = 0.01 m
l = 0.10                  # correlation length: 10 cm = 0.10 m

c = 3e8                   # speed of light (m/s)
theta = np.deg2rad(theta_deg)

# 2. Calculate radar wavenumber
k = 2 * np.pi * frequency / c

# AIEM wave-vector components
kx = k * np.sin(theta)
kz = k * np.cos(theta)

# Required spatial frequency in the AIEM equation
qx = -2 * kx
qy = 0

q = np.sqrt(qx**2 + qy**2)

print("Radar wavenumber k =", k, "rad/m")
print("kx =", kx, "rad/m")
print("kz =", kz, "rad/m")
print("Required q =", q, "rad/m")

# 3. Exponential autocorrelation function
# rho(r) = exp(-r/l)

r = np.linspace(0, 0.5, 300)   # distance in metres
rho = np.exp(-r / l)

# 4. Calculate W^i(qx, qy)
# 2D Fourier transform of [rho(r)]^i
#
# W_i(q) = 2*pi*(i/l) /
#          [((i/l)^2 + q^2)^(3/2)]

orders = np.arange(1, 11)

W_values = []

for i in orders:
    a = i / l

    W_i = (
        2 * np.pi * a
        / ((a**2 + q**2)**1.5)
    )

    W_values.append(W_i)

# 5. Create results table
results = pd.DataFrame({
    "Scattering Order (i)": orders,
    "W_i(-2kx, 0) (m^2)": W_values
})

print("\nAIEM FOURIER SPECTRUM RESULTS")
display(results)

# 6. Plot autocorrelation function
plt.figure(figsize=(7, 5))
plt.plot(r * 100, rho)
plt.xlabel("Distance r (cm)")
plt.ylabel("Autocorrelation rho(r)")
plt.title("Exponential Soil Surface Autocorrelation")
plt.grid(True)
plt.show()

# 7. Plot Fourier spectrum against scattering order
plt.figure(figsize=(7, 5))
plt.plot(orders, W_values, marker="o")
plt.xlabel("Scattering order (i)")
plt.ylabel("W_i(-2kx, 0) (m^2)")
plt.title("AIEM Fourier Spectrum")
plt.grid(True)
plt.show()

# 8. Save results
results.to_csv("Module3_Roughness_Spectrum.csv", index=False)

print("\nModule 3 completed.")
print("Results saved: Module3_Roughness_Spectrum.csv")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from math import factorial

# =====================================================
# AIEM: SCATTERING-ORDER WEIGHTING
# =====================================================

# Reuse values calculated in Module 3
orders = np.asarray(orders, dtype=int)
W_values = np.asarray(W_values, dtype=float)

# RMS roughness in metres
s = 0.01

# Calculate s^(2i) / i! for every scattering order
roughness_weights = np.array([
    s**(2*i) / factorial(i) for i in orders
])

# Calculate the order-dependent factor:
# [s^(2i) / i!] * W_i(-2kx, 0)
order_contributions = roughness_weights * W_values

# Create result table
order_df = pd.DataFrame({
    "Scattering Order (i)": orders,
    "Fourier Spectrum W_i (m^2)": W_values,
    "Roughness Weight s^(2i)/i!": roughness_weights,
    "Order Factor": order_contributions
})

print("AIEM SCATTERING-ORDER WEIGHTING")
display(order_df)

# Plot the order factor
plt.figure(figsize=(7, 5))
plt.semilogy(orders, order_contributions, marker="o")
plt.xlabel("Scattering order (i)")
plt.ylabel("Order factor (log scale)")
plt.title("AIEM Scattering-Order Weighting")
plt.grid(True)
plt.show()

# Save results
order_df.to_csv("AIEM_Scattering_Order_Weights.csv", index=False)

print("Saved: AIEM_Scattering_Order_Weights.csv")
```
