```python
import numpy as np
import matplotlib.pyplot as plt


# 1. FIXED PARAMETERS

theta_deg = 40
lambda_cm = 5.6

theta = np.deg2rad(theta_deg)

cos_theta = np.cos(theta)
sin_theta = np.sin(theta)
cot_theta = 1 / np.tan(theta)

# Radar wavenumber
k = 2 * np.pi / lambda_cm


# 2. VARY SOIL MOISTURE FROM 5 TO 30

SM = np.arange(5, 31, 1)


# 3. SURFACE ROUGHNESS VALUES

roughness_values = [1, 2, 3]   # cm


# 4. PLOT BARE-SOIL BACKSCATTER

plt.figure(figsize=(8, 6))

for s_cm in roughness_values:

    # Dubois Model
    sigma_soil = (
        10**(-1.138)
        * cos_theta**1.528
        * 10**(0.008 * cot_theta * SM)
        * (k * s_cm)**(0.71 * sin_theta)
    )

    # Plot each roughness
    plt.plot(
        SM,
        sigma_soil,
        'o-',
        label=f"Surface roughness = {s_cm} cm"
    )


plt.xlabel("Soil Moisture")
plt.ylabel(r"Bare-soil $\sigma^0_{VV}$")
plt.title(
    "Dubois Model: Effect of Soil Moisture and Surface Roughness"
)

plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
