```python
# ============================================================
# AUTOMATIC MODULATION CLASSIFICATION (AMC)
# BPSK, QPSK, 8PSK, 16QAM + CNN
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D
from tensorflow.keras.layers import GlobalAveragePooling1D, Dense, Dropout

# ------------------------------------------------------------
# 1. SETTINGS
# ------------------------------------------------------------

np.random.seed(42)
tf.random.set_seed(42)

MODULATIONS = ["BPSK", "QPSK", "8PSK", "16QAM"]

NUM_SAMPLES = 128       # I/Q samples in one signal
SAMPLES_PER_CLASS = 1000

print("Modulation classes:", MODULATIONS)


# ------------------------------------------------------------
# 2. GENERATE BPSK
# ------------------------------------------------------------

def generate_bpsk(n):
    bits = np.random.randint(0, 2, n)

    # 0 -> -1
    # 1 -> +1
    symbols = 2 * bits - 1

    return symbols.astype(complex)


# ------------------------------------------------------------
# 3. GENERATE QPSK
# ------------------------------------------------------------

def generate_qpsk(n):

    symbols_index = np.random.randint(0, 4, n)

    phases = np.pi/4 + symbols_index * np.pi/2

    symbols = np.exp(1j * phases)

    return symbols


# ------------------------------------------------------------
# 4. GENERATE 8PSK
# ------------------------------------------------------------

def generate_8psk(n):

    symbols_index = np.random.randint(0, 8, n)

    phases = np.pi/8 + symbols_index * np.pi/4

    symbols = np.exp(1j * phases)

    return symbols


# ------------------------------------------------------------
# 5. GENERATE 16QAM
# ------------------------------------------------------------

def generate_16qam(n):

    levels = np.array([-3, -1, 1, 3])

    I = np.random.choice(levels, n)
    Q = np.random.choice(levels, n)

    symbols = I + 1j * Q

    # Normalization
    symbols = symbols / np.sqrt(10)

    return symbols


# ------------------------------------------------------------
# 6. STORE GENERATORS
# ------------------------------------------------------------

generators = {
    "BPSK": generate_bpsk,
    "QPSK": generate_qpsk,
    "8PSK": generate_8psk,
    "16QAM": generate_16qam
}


# ------------------------------------------------------------
# 7. PLOT CONSTELLATION DIAGRAMS
# ------------------------------------------------------------

plt.figure(figsize=(10, 8))

for i, modulation in enumerate(MODULATIONS):

    signal = generators[modulation](1000)

    plt.subplot(2, 2, i + 1)

    plt.scatter(
        signal.real,
        signal.imag,
        s=8
    )

    plt.title(modulation)
    plt.xlabel("In-phase (I)")
    plt.ylabel("Quadrature (Q)")
    plt.grid()

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 8. FUNCTION TO ADD AWGN NOISE
# ------------------------------------------------------------

def add_awgn(signal, snr_db):

    signal_power = np.mean(np.abs(signal) ** 2)

    noise_power = signal_power / (10 ** (snr_db / 10))

    noise = np.sqrt(noise_power / 2) * (
        np.random.randn(len(signal))
        + 1j * np.random.randn(len(signal))
    )

    noisy_signal = signal + noise

    return noisy_signal


# ------------------------------------------------------------
# 9. CREATE DATASET
# ------------------------------------------------------------

X = []
y = []

print("Generating dataset...")

for label, modulation in enumerate(MODULATIONS):

    for i in range(SAMPLES_PER_CLASS):

        # Generate signal
        signal = generators[modulation](NUM_SAMPLES)

        # Random SNR between 5 and 20 dB
        snr = np.random.uniform(5, 20)

        # Add noise
        signal = add_awgn(signal, snr)

        # Normalize signal
        signal = signal / (
            np.sqrt(np.mean(np.abs(signal) ** 2))
            + 1e-8
        )

        # Store I and Q separately
        iq_signal = np.stack(
            [signal.real, signal.imag],
            axis=-1
        )

        X.append(iq_signal)
        y.append(label)


X = np.array(X)
y = np.array(y)

print("Dataset shape:", X.shape)
print("Labels shape:", y.shape)


# ------------------------------------------------------------
# 10. TRAIN / TEST SPLIT
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining data:", X_train.shape)
print("Testing data:", X_test.shape)


# ------------------------------------------------------------
# 11. BUILD CNN MODEL
# ------------------------------------------------------------

model = Sequential([

    Conv1D(
        32,
        kernel_size=7,
        activation="relu",
        input_shape=(NUM_SAMPLES, 2)
    ),

    MaxPooling1D(pool_size=2),

    Conv1D(
        64,
        kernel_size=5,
        activation="relu"
    ),

    MaxPooling1D(pool_size=2),

    Conv1D(
        128,
        kernel_size=5,
        activation="relu"
    ),

    GlobalAveragePooling1D(),

    Dense(
        64,
        activation="relu"
    ),

    Dropout(0.2),

    Dense(
        len(MODULATIONS),
        activation="softmax"
    )
])


# ------------------------------------------------------------
# 12. SHOW MODEL
# ------------------------------------------------------------

model.summary()


# ------------------------------------------------------------
# 13. COMPILE MODEL
# ------------------------------------------------------------

model.compile(

    optimizer="adam",

    loss="sparse_categorical_crossentropy",

    metrics=["accuracy"]
)


# ------------------------------------------------------------
# 14. TRAIN MODEL
# ------------------------------------------------------------

print("\nTraining CNN...")

history = model.fit(

    X_train,
    y_train,

    epochs=10,

    batch_size=64,

    validation_split=0.2,

    verbose=1
)


# ------------------------------------------------------------
# 15. TEST MODEL
# ------------------------------------------------------------

test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print("\n================================")
print("TEST ACCURACY")
print("================================")

print("Accuracy:", test_accuracy)
print("Accuracy (%):", test_accuracy * 100)


# ------------------------------------------------------------
# 16. PLOT TRAINING ACCURACY
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.title("AMC CNN Training")

plt.legend()
plt.grid()

plt.show()


# ------------------------------------------------------------
# 17. CONFUSION MATRIX
# ------------------------------------------------------------

predictions = model.predict(X_test)

y_pred = np.argmax(
    predictions,
    axis=1
)

cm = confusion_matrix(
    y_test,
    y_pred
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=MODULATIONS
)

disp.plot()

plt.title("AMC Confusion Matrix")

plt.show()


# ------------------------------------------------------------
# 18. TEST AN UNKNOWN SIGNAL
# ------------------------------------------------------------

print("\n================================")
print("TESTING UNKNOWN SIGNAL")
print("================================")

# Let's create an unknown QPSK signal

unknown_signal = generate_qpsk(NUM_SAMPLES)

# Add noise
unknown_signal = add_awgn(
    unknown_signal,
    10
)

# Normalize
unknown_signal = unknown_signal / (
    np.sqrt(
        np.mean(
            np.abs(unknown_signal) ** 2
        )
    )
    + 1e-8
)

# Convert to I/Q
unknown_iq = np.stack(
    [
        unknown_signal.real,
        unknown_signal.imag
    ],
    axis=-1
)

# Add batch dimension
unknown_iq = unknown_iq.reshape(
    1,
    NUM_SAMPLES,
    2
)

# Prediction
prediction = model.predict(
    unknown_iq,
    verbose=0
)

predicted_class = np.argmax(
    prediction
)

confidence = prediction[0][predicted_class]

print(
    "Predicted modulation:",
    MODULATIONS[predicted_class]
)

print(
    "Confidence:",
    confidence * 100,
    "%"
)
```
