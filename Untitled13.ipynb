```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# 1. RADAR PARAMETERS
# ============================================================

theta_deg = 40
lambda_cm = 5.6
s_cm = 1.0

theta = np.deg2rad(theta_deg)

cos_theta = np.cos(theta)
sin_theta = np.sin(theta)

sec_theta = 1 / cos_theta
cot_theta = 1 / np.tan(theta)

k = 2 * np.pi / lambda_cm


# ============================================================
# 2. SIR'S 21-ROW LUT
# ============================================================

lut = pd.DataFrame({

    "LAI": [
        7.49, 7.57, 7.78,
        7.12, 7.22, 6.15,
        7.07, 4.58, 6.11,
        8.77, 8.93, 8.67,
        8.80, 8.26, 8.33,
        6.34, 7.14, 7.47,
        5.66, 5.16, 6.22
    ],

    "SM": [
        0.4694, 0.4694, 0.4694,
        0.2219, 0.2219, 0.2219,
        0.4417, 0.4417, 0.4417,
        0.3413, 0.3413, 0.3413,
        0.4375, 0.4375, 0.4375,
        0.4184, 0.4184, 0.4184,
        0.436666667,
        0.441666667,
        0.435333333
    ],

    "VH_FDP": [
        0.02793, 0.01725, 0.02207,
        0.01524, 0.01381, 0.01683,
        0.02428, 0.01586, 0.02334,
        0.02434, 0.02548, 0.01834,
        0.01996, 0.02899, 0.02539,
        0.02040, 0.02671, 0.03059,
        0.02360, 0.03759, 0.03194
    ],

    "VV_FDP": [
        0.03859, 0.04903, 0.03493,
        0.02832, 0.02317, 0.03057,
        0.04429, 0.02798, 0.03680,
        0.05130, 0.04578, 0.04100,
        0.04121, 0.04840, 0.04711,
        0.05173, 0.03546, 0.04658,
        0.05107, 0.05242, 0.04297
    ]
})


# ============================================================
# 3. DUBOIS MODEL
# ============================================================

def dubois_model_vv(SM):

    # Convert SM from fraction to percentage
    SM_percent = SM * 100

    sigma_soil = (
        10**(-1.138)
        * cos_theta**1.528
        * 10**(
            0.008
            * cot_theta
            * SM_percent
        )
        * (k * s_cm)**(
            0.71 * sin_theta
        )
    )

    return sigma_soil


# Calculate soil backscatter for every LUT row

lut["sigma_soil_VV"] = lut["SM"].apply(
    dubois_model_vv
)


# ============================================================
# 4. OPTIMIZE A AND B
# ============================================================

# We search a wider range because the previous range
# forced the solution to A = 0.01 and B = 0.20.

A_values = np.arange(
    0.001,
    0.0301,
    0.0001
)

B_values = np.arange(
    0.010,
    0.3001,
    0.0001
)


best_rmse = np.inf
best_A = None
best_B = None


# ============================================================
# 5. SEARCH FOR MINIMUM RMSE
# ============================================================

for A in A_values:

    for B in B_values:

        # ----------------------------------------
        # tau squared
        # ----------------------------------------

        tau_squared = np.exp(
            -2
            * B
            * lut["LAI"].values
            * sec_theta
        )


        # ----------------------------------------
        # vegetation backscatter
        # ----------------------------------------

        sigma_veg = (
            A
            * lut["LAI"].values
            * cos_theta
            * (1 - tau_squared)
        )


        # ----------------------------------------
        # total WCM VV
        # ----------------------------------------

        sigma_total = (
            sigma_veg
            + tau_squared
            * lut["sigma_soil_VV"].values
        )


        # ----------------------------------------
        # RMSE using ALL 21 FDP VV values
        # ----------------------------------------

        rmse = np.sqrt(
            np.mean(
                (
                    lut["VV_FDP"].values
                    - sigma_total
                ) ** 2
            )
        )


        # ----------------------------------------
        # Keep the best A and B
        # ----------------------------------------

        if rmse < best_rmse:

            best_rmse = rmse
            best_A = A
            best_B = B


# ============================================================
# 6. PRINT OPTIMIZED A AND B
# ============================================================

print("=" * 60)
print("             OPTIMIZED WCM PARAMETERS")
print("=" * 60)

print(
    f"Optimized A  = {best_A:.6f}"
)

print(
    f"Optimized B  = {best_B:.6f}"
)

print(
    f"Minimum RMSE = {best_rmse:.6f}"
)


# ============================================================
# 7. CALCULATE FINAL WCM USING OPTIMIZED A AND B
# ============================================================

tau_squared = np.exp(
    -2
    * best_B
    * lut["LAI"].values
    * sec_theta
)


# Vegetation contribution

sigma_veg = (
    best_A
    * lut["LAI"].values
    * cos_theta
    * (1 - tau_squared)
)


# Attenuated soil contribution

tau2_sigma_soil = (
    tau_squared
    * lut["sigma_soil_VV"].values
)


# Total WCM backscatter

sigma_total = (
    sigma_veg
    + tau2_sigma_soil
)


# ============================================================
# 8. ADD RESULTS TO LUT
# ============================================================

lut["tau_squared"] = tau_squared

lut["sigma_veg"] = sigma_veg

lut["tau2_sigma_soil"] = tau2_sigma_soil

lut["sigma_vv_simulated"] = sigma_total

lut["error"] = (
    lut["VV_FDP"]
    - lut["sigma_vv_simulated"]
)


# ============================================================
# 9. FINAL RMSE
# ============================================================

final_rmse = np.sqrt(
    np.mean(
        lut["error"] ** 2
    )
)


print()
print("Final RMSE =", final_rmse)


# ============================================================
# 10. SHOW RESULTS
# ============================================================

print()
print("=" * 120)
print("                    FINAL WCM RESULTS")
print("=" * 120)

print(
    lut[
        [
            "LAI",
            "SM",
            "sigma_soil_VV",
            "tau_squared",
            "sigma_veg",
            "tau2_sigma_soil",
            "sigma_vv_simulated",
            "VV_FDP",
            "error"
        ]
    ].to_string(
        index=True,
        float_format=lambda x: f"{x:.6f}"
    )
)


# ============================================================
# 11. OBSERVED VS SIMULATED VV PLOT
# ============================================================

plt.figure(figsize=(7, 7))


# X-axis = WCM simulated
# Y-axis = FDP observed

plt.scatter(
    lut["sigma_vv_simulated"],
    lut["VV_FDP"],
    s=60
)


# 1:1 line

min_value = min(
    lut["sigma_vv_simulated"].min(),
    lut["VV_FDP"].min()
)

max_value = max(
    lut["sigma_vv_simulated"].max(),
    lut["VV_FDP"].max()
)


plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--"
)


# Labels

plt.xlabel(
    r"Simulated $\sigma^0_{VV}$"
)

plt.ylabel(
    r"Observed/FDP $\sigma^0_{VV}$"
)

plt.title(
    "Observed vs Simulated VV"
)


# RMSE on graph

plt.text(
    min_value,
    max_value * 0.95,
    f"RMSE = {final_rmse:.4f}"
)


plt.grid(True)

plt.show()
```
