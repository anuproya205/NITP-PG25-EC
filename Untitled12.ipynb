```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. OBSERVED - AB SINGLE NAHI, PURA LUT USE HOGA
print("Observed VV = All 21 FDP VV values")

# 2. RADAR / SOIL PARAMETERS - same as yours
theta_deg = 40
lambda_cm = 5.6
s_cm = 1.0
theta = np.deg2rad(theta_deg)
cos_theta = np.cos(theta)
sin_theta = np.sin(theta)
sec_theta = 1 / cos_theta
cot_theta = 1 / np.tan(theta)
k = 2 * np.pi / lambda_cm

# 3. FDP LOOK-UP TABLE - same as yours
lut = pd.DataFrame({
    "LAI": [7.49,7.57,7.78,7.12,7.22,6.15,7.07,4.58,6.11,8.77,8.93,8.67,8.80,8.26,8.33,6.34,7.14,7.47,5.66,5.16,6.22],
    "SM": [0.4694,0.4694,0.4694,0.2219,0.2219,0.2219,0.4417,0.4417,0.4417,0.3413,0.3413,0.3413,0.4375,0.4375,0.4375,0.4184,0.4184,0.4184,0.436666667,0.441666667,0.435333333],
    "VV_FDP": [0.03859,0.04903,0.03493,0.02832,0.02317,0.03057,0.04429,0.02798,0.03680,0.05130,0.04578,0.04100,0.04121,0.04840,0.04711,0.05173,0.03546,0.04658,0.05107,0.05242,0.04297]
})

# 4. DUBOIS MODEL - MODIFICATION 1: SM*100 karna tha
def dubois_model_vv(SM):
    SM_percent = SM * 100  # ye aapke code me missing tha
    sigma_soil_vv = (
        10**(-1.138) * cos_theta**1.528 * 10**(0.008 * cot_theta * SM_percent) * (k * s_cm)**(0.71 * sin_theta)
    )
    return sigma_soil_vv

lut["sigma_soil_VV"] = lut["SM"].apply(dubois_model_vv)

# 5. WCM A AND B SEARCH RANGE - MODIFICATION 2: range bada kiya taaki best mile
A_min, A_max, A_step = 0.001, 0.030, 0.0005
B_min, B_max, B_step = 0.010, 0.300, 0.005
A_values = np.arange(A_min, A_max + A_step, A_step)
B_values = np.arange(B_min, B_max + B_step, B_step)

# 6. WCM CALCULATION - MODIFICATION 3: LOGIC SAME, BAS COST SARE 21 SE
# Aapka same triple loop, bas ab global RMSE
best_rmse = np.inf
best_A = None
best_B = None

for A in A_values:
    for B in B_values:
        # Aapka same formula, par 21 rows ke liye ek saath
        tau_squared = np.exp(-2 * B * lut["LAI"].values * sec_theta)
        sigma_veg = A * lut["LAI"].values * cos_theta * (1 - tau_squared)
        sigma_vv_total = sigma_veg + tau_squared * lut["sigma_soil_VV"].values

        # YAHAN BADLAV: sare 21 observed se RMSE, single 0.0287 se nahi
        rmse = np.sqrt(np.mean((lut["VV_FDP"].values - sigma_vv_total)**2))

        if rmse < best_rmse:
            best_rmse = rmse
            best_A = A
            best_B = B

# 7. BEST A,B SE FINAL RESULTS - aapka same results list wala logic
results = []
for row_number, row in lut.iterrows():
    LAI = row["LAI"]
    V1 = LAI
    V2 = LAI
    sigma_soil_vv = row["sigma_soil_VV"]

    tau_squared = np.exp(-2 * best_B * V2 * sec_theta)
    sigma_veg = best_A * V1 * cos_theta * (1 - tau_squared)
    sigma_vv_total = sigma_veg + tau_squared * sigma_soil_vv
    difference = row["VV_FDP"] - sigma_vv_total
    cost = difference**2

    results.append([row_number+1, LAI, row["SM"], sigma_soil_vv, best_A, best_B, tau_squared, sigma_veg, sigma_vv_total, difference, np.sqrt(cost)])

columns = ["LUT_Row","LAI","SM","sigma_soil_VV","A","B","tau_squared","sigma_veg","sigma_vv_total","difference","cost"]
wcm_lut = pd.DataFrame(results, columns=columns)

# 8. FIND MINIMUM COST - same
best_index = wcm_lut["cost"].idxmin()
best_result = wcm_lut.loc[best_index]

# 9. PRINT BEST RESULT - same style
print("\n" + "="*70)
print("      BEST WCM RESULT (21 FDP OPTIMIZED)")
print("="*70)
print(f"Best A                  = {best_A:.8f}")
print(f"Best B                  = {best_B:.8f}")
print(f"Minimum RMSE (21 rows)  = {best_rmse:.15f}")
print(f"Best LAI (min cost row) = {best_result['LAI']:.6f}")
print(f"Best SM                 = {best_result['SM']:.9f}")

# 10. DISPLAY COMPLETE 21-ROW LUT - same
print("\n" + "="*70)
print("          21 FDP LUT ROWS WITH DUBOIS SOIL VV")
print("="*70)
print(lut[["LAI","SM","sigma_soil_VV","VV_FDP"]].to_string(index=True, float_format=lambda x: f"{x:.9f}"))

# 11. DISPLAY 20 LOWEST-COST - same
print("\n" + "="*70)
print("          20 LOWEST-COST WCM COMBINATIONS (BEST A,B)")
print("="*70)
print(wcm_lut.sort_values(by="cost").head(20).to_string(index=False, float_format=lambda x: f"{x:.9f}"))

# ============================================================
# OBSERVED VV vs SIMULATED VV - MODIFICATION 4: BEST A,B USE
# ============================================================
A_plot = best_A
B_plot = best_B
tau_squared_plot = np.exp(-2 * B_plot * lut["LAI"] * sec_theta)
sigma_veg_plot = A_plot * lut["LAI"] * cos_theta * (1 - tau_squared_plot)
sigma_vv_sim = sigma_veg_plot + tau_squared_plot * lut["sigma_soil_VV"]
sigma_vv_obs_lut = lut["VV_FDP"]
rmse_vv = np.sqrt(np.mean((sigma_vv_obs_lut - sigma_vv_sim)**2))

print(f"\nA = {A_plot} B = {B_plot} VV RMSE = {rmse_vv}")

plt.figure(figsize=(7, 7))
plt.scatter(sigma_vv_sim, sigma_vv_obs_lut, s=60)
min_value = min(sigma_vv_sim.min(), sigma_vv_obs_lut.min())
max_value = max(sigma_vv_sim.max(), sigma_vv_obs_lut.max())
plt.plot([min_value, max_value], [min_value, max_value], linestyle='--')
plt.xlabel(r"Simulated $\sigma^0_{VV}$")
plt.ylabel(r"Observed/FDP $\sigma^0_{VV}$")
plt.title(f"Observed vs Simulated VV (Optimized 21 pts)")
plt.text(min_value, max_value * 0.95, f"RMSE = {rmse_vv:.4f}\nA={best_A:.4f} B={best_B:.4f}")
plt.grid(True)
plt.show()
```
