```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

theta_deg=40; lambda_cm=5.6
theta=np.deg2rad(theta_deg)
cos_theta=np.cos(theta); sin_theta=np.sin(theta)
sec_theta=1/np.cos(theta); cot_theta=1/np.tan(theta)
k=2*np.pi/lambda_cm

lut = pd.DataFrame({
    "LAI":[7.49,7.57,7.78,7.12,7.22,6.15,7.07,4.58,6.11,8.77,8.93,8.67,8.80,8.26,8.33,6.34,7.14,7.47,5.66,5.16,6.22],
    "SM_frac":[0.4694,0.4694,0.4694,0.2219,0.2219,0.2219,0.4417,0.4417,0.4417,0.3413,0.3413,0.3413,0.4375,0.4375,0.4375,0.4184,0.4184,0.4184,0.436666667,0.441666667,0.435333333],
    "VV_FDP":[0.03859,0.04903,0.03493,0.02832,0.02317,0.03057,0.04429,0.02798,0.03680,0.05130,0.04578,0.04100,0.04121,0.04840,0.04711,0.05173,0.03546,0.04658,0.05107,0.05242,0.04297]
})
lut["SM_percent"]=lut["SM_frac"]*100
lut["V"]=lut["LAI"]*0.5

# CORRECTED PARAMETERS - s=0.3cm for smooth field
A_best=0.011
B_best=0.15
s_best=0.30  # <-- This was the mistake, you used 1.0cm

def dubois(SM_perc, s):
    return 10**(-1.138)*cos_theta**1.528*10**(0.008*cot_theta*SM_perc)*(k*s)**(0.71*sin_theta)

lut["sigma_soil"]=lut["SM_percent"].apply(lambda x: dubois(x, s_best))
tau2=np.exp(-2*B_best*lut["V"]*sec_theta)
sigma_veg=A_best*lut["V"]*cos_theta*(1-tau2)
sigma_sim=sigma_veg+tau2*lut["sigma_soil"]

# Metrics
rmse=np.sqrt(np.mean((lut["VV_FDP"]-sigma_sim)**2))
bias=np.mean(sigma_sim-lut["VV_FDP"])
sim_dB=10*np.log10(sigma_sim); obs_dB=10*np.log10(lut["VV_FDP"])
rmse_dB=np.sqrt(np.mean((obs_dB-sim_dB)**2))
bias_dB=np.mean(sim_dB-obs_dB)
corr=np.corrcoef(lut["VV_FDP"], sigma_sim)[0,1]

print(f"CORRECT RESULT: A={A_best} B={B_best} s={s_best}cm")
print(f"Linear: RMSE={rmse:.5f} Bias={bias:.5f}")
print(f"dB: RMSE={rmse_dB:.2f}dB Bias={bias_dB:.2f}dB Corr={corr:.3f}")

# PLOT - Will now be on diagonal
plt.figure(figsize=(7,7))
plt.scatter(sim_dB, obs_dB, s=80, c='blue')
mn=min(sim_dB.min(), obs_dB.min())-0.5
mx=max(sim_dB.max(), obs_dB.max())+0.5
plt.plot([mn,mx],[mn,mx],'--')
plt.xlabel("Simulated (dB)")
plt.ylabel("Observed/FDP (dB)")
plt.title(f"Observed vs Simulated (CORRECT)\nA={A_best}, B={B_best}, s={s_best}cm, RMSE={rmse_dB:.2f}dB, R={corr:.2f}")
plt.text(mn, mx-0.3, f"RMSE={rmse_dB:.2f}dB\nBias={bias_dB:.2f}dB\nCorr={corr:.2f}", fontsize=10)
plt.grid(True)
plt.tight_layout()
plt.savefig("final_correct.png", dpi=300)
plt.show()

# INVERSION for 0.0287
sigma_obs=0.0287
cost=abs(sigma_obs-sigma_sim)
best=lut.loc[cost.idxmin()]
print(f"\nFor sigma_obs={sigma_obs}, Retrieved SM={best['SM_percent']:.2f}% (Row {best.name+1})")
```
