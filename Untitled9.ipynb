```python
import numpy as np

# -----------------------------
# FIXED INPUTS
# -----------------------------

theta_deg = 40       # incidence angle in degrees
lambda_cm = 5.6      # wavelength in cm
epsilon = 8.5        # real part of dielectric constant
s_cm = 1.0           # RMS soil roughness in cm

# Convert angle from degrees to radians
theta = np.deg2rad(theta_deg)

# Radar wavenumber
k = 2 * np.pi / lambda_cm

# -----------------------------
# DUBOIS VV MODEL
# -----------------------------

sigma_soil_vv = (
    10**(-2.35)
    * (np.cos(theta)**3 / np.sin(theta)**3)
    * 10**(0.046 * epsilon * np.tan(theta))
    * (k * s_cm * np.sin(theta))**1.1
    * lambda_cm**0.7
)

print("Radar wavenumber k =", k)
print("Soil VV backscatter (linear) =", sigma_soil_vv)

# Convert to dB
sigma_soil_vv_dB = 10 * np.log10(sigma_soil_vv)

print("Soil VV backscatter (dB) =", sigma_soil_vv_dB)
```

```python
import numpy as np
import pandas as pd

# ============================================================
# STEP 2 : WATER CLOUD MODEL (WCM)
# ============================================================


# ============================================================
# 1. FIXED INPUTS
# ============================================================

# Observed VV backscatter
sigma_vv_obs = 0.0287

# Incidence angle
theta_deg = 40

# Convert degrees to radians
theta = np.deg2rad(theta_deg)

# Calculate cosine and secant
cos_theta = np.cos(theta)
sec_theta = 1 / cos_theta


# ============================================================
# 2. SOIL BACKSCATTER FROM DUBOIS (STEP 1)
# ============================================================

# Result obtained from your Step-1 Dubois calculation
sigma_soil_vv = 0.037518395397274036

print("Soil VV backscatter from Dubois =", sigma_soil_vv)


# ============================================================
# 3. LAI VALUES FROM YOUR LUT
# ============================================================

LAI_values = np.array([
    7.49,
    7.57,
    7.78,
    7.12,
    7.22,
    6.15,
    7.07,
    4.58,
    6.11,
    8.77,
    8.93,
    8.67,
    8.80,
    8.26,
    8.33,
    6.34,
    7.14,
    7.47,
    5.66,
    5.16,
    6.22
])


# ============================================================
# 4. A AND B VALUES
# ============================================================

# TEMPORARY SEARCH RANGES
# These must be replaced by the ranges specified by your sir
# or the WCM reference.

A_values = np.arange(0.001, 0.101, 0.001)

B_values = np.arange(0.01, 1.01, 0.01)


# ============================================================
# 5. EMPTY LIST TO STORE ALL LUT RESULTS
# ============================================================

results = []


# ============================================================
# 6. LOOP THROUGH LAI
# ============================================================

for LAI in LAI_values:

    # Sir said:
    # V1 = V2 = LAI

    V1 = LAI
    V2 = LAI


    # ========================================================
    # 7. LOOP THROUGH A
    # ========================================================

    for A in A_values:


        # ====================================================
        # 8. LOOP THROUGH B
        # ====================================================

        for B in B_values:


            # ------------------------------------------------
            # WCM EQUATION 1
            #
            # tau² = exp(-2 B V2 sec(theta))
            # ------------------------------------------------

            tau_squared = np.exp(
                -2 * B * V2 * sec_theta
            )


            # ------------------------------------------------
            # WCM EQUATION 2
            #
            # sigma_veg =
            # A V1 cos(theta) (1 - tau²)
            # ------------------------------------------------

            sigma_veg = (
                A
                * V1
                * cos_theta
                * (1 - tau_squared)
            )


            # ------------------------------------------------
            # WCM EQUATION 3
            #
            # sigma_total =
            # sigma_veg + tau² sigma_soil
            # ------------------------------------------------

            sigma_vv_total = (
                sigma_veg
                + tau_squared * sigma_soil_vv
            )


            # ------------------------------------------------
            # ERROR
            #
            # Observed - calculated
            # ------------------------------------------------

            difference = (
                sigma_vv_obs - sigma_vv_total
            )


            # ------------------------------------------------
            # COST FUNCTION
            #
            # Only ONE observed VV value is being used.
            # Therefore RMSE = absolute error.
            # ------------------------------------------------

            cost = np.sqrt(
                difference ** 2
            )


            # ------------------------------------------------
            # SAVE RESULT
            # ------------------------------------------------

            results.append([
                LAI,
                A,
                B,
                tau_squared,
                sigma_veg,
                sigma_vv_total,
                difference,
                cost
            ])


# ============================================================
# 9. CREATE LUT DATAFRAME
# ============================================================

columns = [
    "LAI",
    "A",
    "B",
    "tau_squared",
    "sigma_veg",
    "sigma_vv_total",
    "difference",
    "cost"
]

lut_results = pd.DataFrame(
    results,
    columns=columns
)


# ============================================================
# 10. FIND MINIMUM COST
# ============================================================

best_index = lut_results["cost"].idxmin()

best_result = lut_results.loc[best_index]


# ============================================================
# 11. DISPLAY BEST RESULT
# ============================================================

print("\n==============================================")
print("              BEST WCM RESULT")
print("==============================================")

print(f"Observed VV             = {sigma_vv_obs:.15f}")
print(f"Soil VV from Dubois     = {sigma_soil_vv:.15f}")

print(f"Best LAI                = {best_result['LAI']:.2f}")
print(f"Best A                  = {best_result['A']:.6f}")
print(f"Best B                  = {best_result['B']:.6f}")

print(f"Best tau²               = {best_result['tau_squared']:.15f}")
print(f"Vegetation VV           = {best_result['sigma_veg']:.15f}")

print(f"Calculated Total VV     = {best_result['sigma_vv_total']:.15f}")

print(f"Difference              = {best_result['difference']:.15f}")
print(f"Minimum Cost            = {best_result['cost']:.15f}")


# ============================================================
# 12. SHOW 10 LOWEST-COST COMBINATIONS
# ============================================================

best_10 = lut_results.sort_values(
    by="cost"
).head(10)

print("\n==============================================")
print("       10 LOWEST-COST LUT COMBINATIONS")
print("==============================================")

print(
    best_10[
        [
            "LAI",
            "A",
            "B",
            "tau_squared",
            "sigma_veg",
            "sigma_vv_total",
            "difference",
            "cost"
        ]
    ].to_string(
        index=False,
        float_format=lambda x: f"{x:.15f}"
    )
)
```
