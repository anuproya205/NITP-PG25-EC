```python
import numpy as np

def dubois_vv(sm, theta_deg=30, s_cm=1.0):
    # Topp: SM -> epsilon
    eps = 3.03 + 9.3*sm + 146*sm**2 - 76.7*sm**3
    theta = np.radians(theta_deg)
    lam_cm = 5.54 # Sentinel-1 C-band
    k = 2*np.pi / (lam_cm/100)
    s = s_cm/100
    k_s_sin = k*s*np.sin(theta)

    # Your screenshot Eq (3) - VV only
    sigma_vv = (10**-2.35 * (np.cos(theta)**3 / np.sin(theta)**3) *
                10**(0.046*eps*np.tan(theta)) *
                (k_s_sin)**1.1 * lam_cm**0.7)
    return sigma_vv

# Moisture change karo 0.05 se 0.5
for sm in [0.1, 0.2, 0.3, 0.4]:
    vv_linear = dubois_vv(sm)
    vv_db = 10*np.log10(vv_linear)
    print(f"SM={sm} -> Sigma Soil VV = {vv_db:.2f} dB")
```

```python
import numpy as np
import pandas as pd

# -----------------------------
# FIXED INPUTS
# -----------------------------
theta_deg = 30       # Sir 30 lete hain, tum 40 le rahe the
lambda_cm = 5.54     # S1 = 5.54, tum 5.6 le rahe the
s_cm = 1.0
theta = np.deg2rad(theta_deg)
sec_theta = 1 / np.cos(theta)

# Observed
sigma_vv_obs = 0.0287 # Sir ki slide se

# LAI + mv ki LUT - Sir ki photo wali
LUT_input = np.array([
    [7.49, 0.4694],
    [7.57, 0.4347],
    [7.78, 0.4559],
    [7.12, 0.4421],
    [4.58, 0.4417], # ye best niklega
    [6.22, 0.4353],
])

def epsilon_from_mv(mv):
    return 3.03 + 9.3*mv + 146*mv**2 - 76.7*mv**3

def sigma_soil_dubois_vv(mv, theta, s_cm, lambda_cm):
    eps = epsilon_from_mv(mv)
    k = 2 * np.pi / lambda_cm # cm-1, kyuki s bhi cm me
    sigma = (10**-2.35 * (np.cos(theta)**3 / np.sin(theta)**3) *
             10**(0.046 * eps * np.tan(theta)) *
             (k * s_cm * np.sin(theta))**1.1 * lambda_cm**0.7)
    return sigma, eps

# A,B search - tumhara range thik hai
A_values = np.arange(0.01, 0.5, 0.01)
B_values = np.arange(0.01, 0.5, 0.01)

results = []
for LAI, mv in LUT_input:
    sigma_soil_vv, eps = sigma_soil_dubois_vv(mv, theta, s_cm, lambda_cm)

    for A in A_values:
        for B in B_values:
            tau_squared = np.exp(-2 * B * LAI * sec_theta)
            sigma_veg = A * LAI * np.cos(theta) * (1 - tau_squared)
            sigma_vv_total = sigma_veg + tau_squared * sigma_soil_vv

            cost = abs(sigma_vv_obs - sigma_vv_total) # VV only ke liye bas abs

            results.append([LAI, mv, eps, sigma_soil_vv, A, B, tau_squared, sigma_veg, sigma_vv_total, cost])

columns = ["LAI","mv","epsilon","sigma_soil_vv","A","B","tau2","sigma_veg","sigma_total","cost"]
df = pd.DataFrame(results, columns=columns)

best = df.loc[df["cost"].idxmin()]

print("BEST RESULT")
print(f"Observed VV = {sigma_vv_obs}")
print(f"Best LAI = {best['LAI']}, Best mv = {best['mv']}")
print(f"Epsilon = {best['epsilon']:.2f}")
print(f"Soil VV = {best['sigma_soil_vv']:.5f}")
print(f"Best A = {best['A']}, Best B = {best['B']}")
print(f"Total VV = {best['sigma_total']:.5f}, Cost = {best['cost']:.6f}")

print("\nTop 10:")
print(df.sort_values("cost").head(10)[["LAI","mv","A","B","sigma_total","cost"]].to_string(index=False))
```
